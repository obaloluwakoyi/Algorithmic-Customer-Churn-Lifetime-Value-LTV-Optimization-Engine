```python
import pandas as pd

#  Load the Uplift Output file
file_path = r"C:\Users\HP\Desktop\Data science 12\Uplift_Optimized_Campaign.csv"
df = pd.read_csv(file_path)

```

```python
#  Filter explicitly for Persuadables (TARGET)
persuadables_df = df[df['uplift_segment'] == 'Persuadables (TARGET)']

#  Sort by highest expected net return (net EV)
persuadables_sorted = persuadables_df.sort_values(by='uplift_net_ev', ascending=False)

#  Save actionable customer list for the marketing team
export_path = r"C:\Users\HP\Desktop\Data science 12\High_Priority_Persuadables_List.csv"
persuadables_sorted.to_csv(export_path, index=False)

print(f"Successfully extracted {len(persuadables_sorted)} Persuadable accounts!")
print(f"File exported to: {export_path}\n")

# Display top 10 accounts to target
cols_to_show = ['Geography_Germany', 'Gender_Male', 'Age', 'Balance', 'churn_probability', 'uplift_score', 'uplift_net_ev']
print(persuadables_sorted[cols_to_show].head(10))
```
